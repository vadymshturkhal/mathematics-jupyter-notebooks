\text{How many of the numbers } 2^m, 
\text{ for } 0 \le m \le M, 
\text{ have leading digit 1 in decimal notation? }
$$

```python
M = 37
iterative_sum = 0
print("m, 2**m")
for m in range(M+1):
    if f"{2**m}"[0] == '1':
        print(f"{m}, {2**m}")
        iterative_sum += 1
print()
print(f"{iterative_sum=}")
```
