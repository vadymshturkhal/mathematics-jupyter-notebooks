## $$
\text{Find the sum of all multiples of } (x) \text{ in the closed interval } [a . . b], \text{ when } x > 0
$$

```python
x = 2
a = 50
b = 52

iterative_sum = sum([x*k for k in range(a, b+1)])
print(f"{iterative_sum = }")
print(f"{2*50 + 2*51 + 2*52 = }" )
```
