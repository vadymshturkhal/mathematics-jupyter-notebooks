### Use a summation factor to solve the recurrence

### $T_0 = 5;$

### $2T_n = nT_{n-1} + 3n!$ , for n > 0.

### $T_n = 3n! + \frac{n!}{2^{n-1}}$

```python
from math import factorial


def T_recurrence(n: int):
    if n == 0:
        return 5
    
    Tn = n*T_recurrence(n-1) + 3*factorial(n)
    
    return Tn / 2
```

```python
interval = 12
closed_T = lambda n: 3*factorial(n) + factorial(n)/2**(n-1)
for n in range(interval):
    print(f"{n = }, {T_recurrence(n) = }, {closed_T(n)}")
```

### General form: 

### $a_nT_n = b_nT_{n-1} + c_n$

### The idea is to multiply both sides by summation factor: $s_n$

### $ s_na_nT_n = s_nb_nT_{n-1} + s_nc_n$

# $s_n =  \frac{a_{n-1}a_{n-2}...{a_1}}{b_nb_{n-1}...b_2}$

## $a_n = 2;$

## $b_n = n;$

## $ s_n = \frac{2^{n-1}}{n!}$

### $2T_n = nT_{n-1} + 3n! | * \frac{2^{n-1}}{n!} => $

### $\frac{2^{n}}{n!}T_n = \frac{n2^{n-1}}{n!}T_{n-1} + 3*2^{n-1}!$

## $S_n = s_na_nT_n => S_n = S_{n-1} + s_nc_n$

### $ S_n = s_1b_1T_0 + \sum_{k=1}^{n} s_kc_k$

### $ S_n = 1*1*5 + \sum_{k=1}^{n} \frac{2^{k-1}}{k!}*3k!$

### $ S_n = 5 + \sum_{k=1}^{n} 3*2^{k-1}$

### $ S_n = 5 + 3\sum_{k=1}^{n} 2^{k-1}$

### $ S_n = 5 + 3*(2^{n}-1) $

### $ S_n = 3*2^{n} + 2 $

## $ T_n = \frac{1}{s_na_n}S_n$

## $ T_n = \frac{1}{\frac{2^{n-1}}{n!}*2}(3*2^{n} + 2)$

## $ T_n = \frac{1}{\frac{2^{n}}{n!}}(3*2^{n} + 2)$

## $ T_n = \frac{n!}{2^{n}}(3*2^{n} + 2)$

## $ T_n = 3n! + \frac{n!}{2^{n-1}}$

```python
n = 2
equation = 5 + sum([3*2**k for k in range(n)])
print(equation)
print(5 + 3*(2**(n) - 1))
Sn = 3*2**(n) + 2
print(Sn)

sn = 2**(n-1)/factorial(n)
an = 2
closed_form = Sn/(sn*an)
print(closed_form)
closed_Tn = 3*factorial(n) + factorial(n)/2**(n-1)
print(closed_Tn)
```
