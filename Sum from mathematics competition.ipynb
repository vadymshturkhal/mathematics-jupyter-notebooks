## $$
\text{Evaluate the sum from mathematics competition}\\ 
\frac{1/2}{1+\sqrt{2}}+\frac{1/4}{1+\sqrt[4]{2}}+\frac{1/8}{1+\sqrt[8]{2}}+\frac{1/16}{1+\sqrt[16]{2}}+\cdots\\
\text{from } 1 \text{ to } n\\ 
$$

## $$
\text{First rewrite the sum using Sigma notation}
$$

## $$
\sum\limits_{1\le k \le n} \frac{(1/2)^k}{1 + 2^{\frac{1}{2^k}}}
$$

## <center> Check values of the sum for small n

```python
def mathematics_competition_sum(n: int):
    if n <= 0:
        return 0
    
    return sum([(1/2)**k / (1 + 2**((1/2)**k)) for k in range(1, n+1)])

def mathematics_competition_sum_recurrence(n: int):
    if n <= 0:
        return 0
    
    return mathematics_competition_sum_recurrence(n-1) + (1/2)**n / (1 + 2**((1/2)**n))

closed_form_mathematics_competition_sum = lambda n: -1 - (1/2**n)/(1-2**((1/2)**n))
```

```python
n = 3
print(f"{mathematics_competition_sum(n)=}")
print(f"{mathematics_competition_sum_recurrence(n)=}")
print(f"{closed_form_mathematics_competition_sum(n)=}")
```
