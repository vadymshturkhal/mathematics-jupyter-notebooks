### Use the repertoire method to find a closed form for $\sum_{k=1}^{n} (-1)^{k}k^2$

Repertoire method: $R_n = A(n)a + B(n)b + C(n)c + G(n)g$

Repertoire method: $R_n = R_{n-1} + b + cn + gn^2$

$R_n = R_{n-1} + (-1)^n(b + cn + gn^2)$

Recurrence: $R_n = R_{n-1} + (-1)^nn^2$

```python
def recursive_sum(n: int):
    if n <= 0:
        return 0
    return recursive_sum(n-1) + (-1)**n*n**2
```

```python
A = lambda n: 1
B = lambda n:((-1)**n - 1)/2
C = lambda n: ((-1)**n*n + B(n)) / 2
G = lambda n: ((-1)**n*n**2 - B(n) + 2*C(n)) / 2
```

```python
n = 6334
print(sum([(-1)**k*k**2 for k in range(1, n+1)]))
# print(recursive_sum(n))
closed_form = G(n)
print(closed_form)
```
