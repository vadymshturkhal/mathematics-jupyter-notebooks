# Using a summation factor to solve The Tower of Hanoi recurrence

# $T_0 = 0$

# $T_n = 2T_{n-1} + 1$ , for n > 0

# The Tower of Hanoi recurrence solution: $T_n = 2^n  - 1$

### General form: 

### $a_nT_n = b_nT_{n-1} + c_n$

### The idea is to multiply both sides by summation factor: $s_n$

### $ s_na_nT_n = s_nb_nT_{n-1} + s_nc_n$

# $s_n =  \frac{a_{n-1}a_{n-2}...{a_1}}{b_nb_{n-1}...b_2}$

## $a_n = 1, b_n = 2, c_n = 1, s_n = \frac{1}{2^n}$

```python
def hanoi_recurrence(n: int):
    if n <= 0:
        return 0
    
    return 2*hanoi_recurrence(n-1) + 1
```

```python
n = 7
print(hanoi_recurrence(n))
```

### $ S_n = s_1b_1T_0 + \sum_{k=1}^{n} s_kc_k$

### $ S_n = \frac{1}{2}\cdot2\cdot0 + \sum_{k=1}^{n} \frac{1}{2^k}\cdot1$

### $ S_n = \sum_{k=1}^{n} \frac{1}{2^k}$

### $ S_n = \sum_{k=1}^{n} 2^{-k}$

# Solving $ S_n = \sum_{k=1}^{n} \frac{1}{2^k}$ using Perturbation method

# Closed form: $ S_n = 1 - \frac{1}{2^{n}}$ 

## $ S_n + \frac{1}{2^{n+1}}= \frac{1}{2} + \sum_{k=1}^{n} \frac{1}{2^{k + 1}}$

## $\frac{1}{2} + \sum_{k=1}^{n} \frac{1}{2^{k + 1}}$

## $\frac{1}{2} + \sum_{k=1}^{n} \frac{1}{2^k\cdot2}$

## $\frac{1}{2} + \sum_{k=1}^{n} \frac{1}{2^k}\cdot \frac{1}{2}$

## $\frac{1}{2} + \frac{1}{2} \sum_{k=1}^{n} \frac{1}{2^k}$

## $\frac{1}{2} + \frac{1}{2} S_n$

## $ S_n + \frac{1}{2^{n+1}}= \frac{1}{2} + \frac{1}{2} S_n$

## $ S_n - \frac{1}{2} S_n = \frac{1}{2} - \frac{1}{2^{n+1}}$

## $ \frac{1}{2} S_n = \frac{1}{2} - \frac{1}{2^{n+1}} | \cdot 2$ 

## $ S_n = 1 - \frac{1}{2^{n}}$ 

# $ T_n = \frac{1}{s_n} \cdot S_n $

# $T_n = \frac{1}{\frac{1}{2^n}} \cdot (1 - \frac{1}{2^{n}})$

# $T_n = 2^n \cdot (1 - \frac{1}{2^{n}})$

# $T_n = 2^n  - 1$

```python
S = lambda n: sum([(1/2**k) for k in range(1, n + 1)])
```
