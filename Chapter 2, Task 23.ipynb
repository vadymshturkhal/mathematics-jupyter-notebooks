# $$
\text{Evaluate the sum} \sum\nolimits_{k=1}^{n} \frac{2k+1}{k(k+1)} \text{ in two ways:} \\
\text{a) Replace } \biggl( \frac{1}{k(k+1)} \biggl) \text{ by the "partial fractions" } \biggl( \frac{1}{k} - \frac{1}{k+1} \biggl) ; \\
\text{b) Sum by parts.}
$$

```python
n = 8
iterative_sum = sum([(2*k+1)/(k*(k+1)) for k in range(1, n+1)])
print(f"{iterative_sum=}")
```

# <center> a)

## $$
\sum\nolimits_{k=1}^{n} (2k + 1) \biggl( \frac{1}{k} - \frac{1}{k+1} \biggl)
$$

## $$
\sum\nolimits_{k=1}^{n} \biggl( \frac{2k}{k} - \frac{2k}{k+1} + \frac{1}{k} - \frac{1}{k+1} \biggl)
$$

```python
n = 8
iterative_sum = sum([((2*k)/k) - (2*k)/(k+1) + 1/k - 1/(k+1) for k in range(1, n+1)])
print(f"{iterative_sum=}")
```

## $$
\sum\nolimits_{k=1}^{n} \frac{2k}{k} - \sum\nolimits_{k=1}^{n} \frac{2k}{k+1} + \sum\nolimits_{k=1}^{n} \frac{1}{k} - \sum\nolimits_{k=1}^{n} \frac{1}{k+1}
$$

## $$
\sum\nolimits_{k=1}^{n} 2 - 2\sum\nolimits_{k=1}^{n} \frac{k}{k+1} + H_n - \sum\nolimits_{k=1}^{n} \frac{1}{k+1}
$$

## $$
\text{Harmonic number } H_n = \sum\nolimits_{k=1}^{n} \frac{1}{k}
$$



## $$ 
\sum\limits_{1 \le k \le n} \frac{1}{k+1} = \sum\limits_{1 \le k-1 \le n} \frac{1}{k} = \sum\limits_{2 \le k \le n+1} \frac{1}{k} = H_n - 1 + \frac{1}{n+1}
$$

```python
n = 8
H = lambda n: sum([1/k for k in range(1, n + 1)])
print(f"{H(n)=}")
print()

fourth_sum = sum([1/(k+1) for k in range(1, n + 1)])
print(f"{fourth_sum=}")
print(f"{H(n) - 1 + 1/(n+1)=}")
```

## $$
2n - 2\sum\nolimits_{k=1}^{n} \frac{k}{k+1} + H_n - \bigl(H_n - 1 + \frac{1}{n+1}\bigl)
$$

## <center> Consider second sum

## $$
\frac{k}{k+1} = \frac{k + 1 - 1}{k+1} = \frac{k+1}{k+1} - \frac{1}{k+1} = 1 - \frac{1}{k+1}
$$

## $$
2\sum\nolimits_{k=1}^{n} \bigl(1 - \frac{1}{k+1}\bigl)
$$

## $$
2\sum\nolimits_{k=1}^{n} 1 - 2\sum\nolimits_{k=1}^{n}\frac{1}{k+1} = 2n - 2\bigl(H_n - 1 + \frac{1}{n+1} \bigl) = 2n - 2H_n + 2 - \frac{2}{n+1}
$$

## <center> Return to main sum

## $$
2n - (2n - 2H_n + 2 - \frac{2}{n+1}) + H_n - \bigl(H_n - 1 + \frac{1}{n+1}\bigl)
$$

## $$
2n - 2n + 2H_n - 2 + \frac{2}{n+1} + H_n - H_n + 1 - \frac{1}{n+1}
$$

## $$
2H_n - 1 + \frac{1}{n+1}
$$

```python
n = 8
iterative_sum = sum([((2*k)/k) - (2*k)/(k+1) + 1/k - 1/(k+1) for k in range(1, n+1)])
print(f"{iterative_sum=}")

closed_form = 2*H(n) - 1 + 1/(n+1)
print(f"{closed_form=}")
```

## <center> Falling powers?

# $ x^{\underline{2}} = x(x-1)$ 

## $ x^{\underline{-2}} = \frac{1}{(x+1)(x+2)} $

## <center> Consider the second term

## $$
\sum\limits_{1 \le k \le n} \frac{1}{k(k+1)}
$$

## $$
\sum\limits_{1 \le k+1 \le n} \frac{1}{(k+1)(k+2)}
$$

## $$
\sum\limits_{0 \le k < n} \frac{1}{(k+1)(k+2)}
$$

```python
n = 13
print(sum([1/(k*(k+1)) for k in range(1, n+1)]))
print(sum([1/((k+1)*(k+2)) for k in range(n)]))
```

## <center> Consider the first term

## $$
2\sum\limits_{1 \le k \le n} \frac{k}{k(k+1)}
$$

## $$
2\sum\limits_{1 \le k+1 \le n} \frac{k+1}{(k+1)(k+2)}
$$

## $$
2\sum\limits_{0 \le k < n} \frac{k+1}{(k+1)(k+2)}
$$

## <center> Return to the main sum and rewrite

## $$
2\sum\limits_{0 \le k < n} \frac{k+1}{(k+1)(k+2)} + \sum\limits_{0 \le k < n} \frac{1}{(k+1)(k+2)}
$$

## $$
2\sum\limits_{0 \le k < n} \biggl( \frac{k}{(k+1)(k+2)} + \frac{1}{(k+1)(k+2)} \biggl) + \sum\limits_{0 \le k < n} k^{\underline{-2}}
$$

## $$
2\sum\limits_{0 \le k < n} \frac{k}{(k+1)(k+2)} + 2\sum\limits_{0 \le k < n} \frac{1}{(k+1)(k+2)} + \sum\limits_{0 \le k < n} k^{\underline{-2}}
$$

## $$
2\sum\limits_{0 \le k < n} k \cdot k^{\underline{-2}} + 2\sum\limits_{0 \le k < n} k^{\underline{-2}} + \sum\limits_{0 \le k < n} k^{\underline{-2}}
$$

## $$
2\sum\limits_{0 \le k < n} k \cdot k^{\underline{-2}} + 3\sum\limits_{0 \le k < n} k^{\underline{-2}}
$$

```python
n = 8
first_sum = 2*sum([k*(1/((k+1)*(k+2))) for k in range(n)])
second_sum = 3*sum([1/((k+1)*(k+2)) for k in range(n)])

iterative_sum = first_sum + second_sum
print(f"{iterative_sum=}")

closed_form = 2*H(n) - 1 + 1/(n+1)
print(f"{closed_form=}")
```
