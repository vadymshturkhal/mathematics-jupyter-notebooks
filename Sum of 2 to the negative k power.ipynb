## $$
\text { Solve } \sum\limits_{0 \le k < n} \frac{1}{2^k} \text{ using integral }
$$

## Our strategy is to: Calculate the integral, Calculate Error and we are done!

## $$ \text{Calculate value of} \int_{0}^{n} \frac{1}{2^k} \,dx $$

## $$ \int_{0}^{n} \frac{1}{2^x} \,dx = \frac{2^{-x}}{ln(1/2)}$$

## $$ \frac{2^{-n}}{ln(1/2)} - \frac{2^{0}}{ln(1/2)} = \frac{1-2^{-n}}{ln(2)}$$

```python
from math import log

n = 8
sum_of_2_to_the_negative_k = sum([(1/2**k) for k in range(n)])
print(sum_of_2_to_the_negative_k)

integral = (1 - 2**(-n))/log(2)
print(integral)
```

## <center> What is the difference between Summation and Integral values (what is the error)?

## $$
E_n = S_n - Integral\\
E_n \text{ is the error }\\
S_n = \sum\limits_{1 \le k < n} \frac{1}{2^k}
$$

## $$
E_n = S_n - \frac{1-2^{-n}}{ln(2)}
$$

## <center> Calculate $E_n$

## $$ S_n - \int_{0}^{n} 2^{-x} \,dx = 1 + \sum\limits_{k = 1}^{n} \Bigl( \frac{1}{2^k} - \int_{k-1}^{k} \frac{1}{2^x}dx \Bigl) $$

## $$ 1 + \sum\limits_{k = 1}^{n} \Bigl( \frac{1}{2^k} - \frac{(1-2^{-k}) - (1 - 2^{-(k-1)})}{ln(2)} \Bigl) $$

## $$ 1 + \sum\limits_{k = 1}^{n} \Bigl( \frac{1}{2^k} - \frac{-2^{-k} + 2^{-(k-1)})}{ln(2)} \Bigl) $$

## $$ 1 + \sum\limits_{k = 1}^{n} \Bigl( \frac{1}{2^k} - \frac{2^{-(k-1)} - 2^{-k} }{ln(2)} \Bigl) $$

## $$ 1 + \sum\limits_{k = 1}^{n} \Bigl( 2^{-k} - \frac{2^{-(k-1)} - 2^{-k} }{ln(2)} \Bigl) $$

## $$ 1 + \sum\limits_{k = 1}^{n} \Bigl( 2^{-k} - \frac{2^{-k}}{ln(2)} \Bigl) $$

## $$ 1 + \sum\limits_{k = 1}^{n} 2^{-k}\bigl(1 - \frac{1}{ln(2)} \bigl) $$

## $$ 1 + \bigl(1 - \frac{1}{ln(2)} \bigl)\sum\limits_{k = 1}^{n} 2^{-k} $$

## $$ 1 + \bigl(1 - \frac{1}{ln(2)} \bigl)S_n $$

## $$ S_n - \int_{0}^{n} 2^{-x} \,dx = 1 + \bigl(1 - \frac{1}{ln(2)} \bigl)S_n $$

## $$ S_n - \frac{1-2^{-n}}{ln(2)} = 1 + \bigl(1 - \frac{1}{ln(2)} \bigl)S_n $$

## $$ S_n - \bigl(1 - \frac{1}{ln(2)} \bigl)S_n = 1 + \frac{1-2^{-n}}{ln(2)} $$

## $$ S_n (1 - \bigl(1 - \frac{1}{ln(2)} \bigl) = 1 + \frac{1-2^{-n}}{ln(2)} $$

## $$ S_n (1 - 1 + \frac{1}{ln(2)}) = 1 + \frac{1-2^{-n}}{ln(2)} $$

## $$ S_n \frac{1}{ln(2)} = \frac{1-2^{-n}}{ln(2)} $$

## $$ S_n = \frac{1-2^{-n}}{ln(2)} \cdot ln(2) $$

## $$ \text{Closed form } S_n = \frac{1-2^{-n}}{ln(2)} \cdot ln(2) + 1 = 1-2^{-n} + 1 $$

```python
def error(n: int):
    if n < 0:
        return 0
    
    sum_ = 1
    
    for k in range(n):
        sum_ += 2**-k - (2**-(k-1) - 2**-k)/log(2)
        
    return sum_

interval = 18

print("n", "sum_of_2_to_the_negative_k", "closed_form")
for n in range(interval):
    sum_of_2_to_the_negative_k = sum([(2**-k) for k in range(n)])
    closed_form_sn = (1-2**-(n-1)) + 1
    print(n, sum_of_2_to_the_negative_k, closed_form_sn)
```
