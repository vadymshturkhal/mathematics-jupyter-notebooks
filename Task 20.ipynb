## <center> Try to evaluate $\sum_{k=0}^{n} kH_k$ by the perturbation method, </center>

## <center> but deduce the value of $\sum_{k=0}^{n} H_k$ instead </center>

# <center> Closed form $ \sum_{k=0}^{n} H_{k} = H_{n+1}(n+1) - n - 1 $

## <center> $ S_n = \sum_{k=0}^{n} kH_k $

##  <center> "Harmonic sum" = $ H_k = \sum_{k=1}^{n} \frac{1}{k} $

```python
n = 2

def harmonic_sum(n: int):
    if n <= 0:
        return 0
    
    return sum([1/k for k in range(1, n+1)])
    
    
Sn = sum([k*harmonic_sum(k) for k in range(n+1)])
print(f"With {n=}: {Sn=}")
```

## <center> Solving $n=2$ case </center>

## <center> $ S_2 = 0 + 1H_1 + 2H_2 $

## <center> $ H_1 = \frac{1}{1} = 1 $

## <center> $ H_2 = \frac{1}{1} + \frac{1}{2} = 1.5 $

## <center> $ S_2 = 0 + 1 + 3 = 4 $

## Returning to perturbation method $\dots$

## $ Sn = \sum_{k=0}^{n} kH_k $

## $ Sn + (n+1)H_{n+1} = 0 + \sum_{k=0}^{n} (k+1)H_{k+1} $

```python
print(Sn + (n+1)*harmonic_sum(n+1))
print(sum([(k+1)*harmonic_sum(k+1) for k in range(n+1)]))
```

## $ Sn + (n+1)H_{n+1} = \sum_{k=0}^{n} (k+1)H_{k+1} $

## $ Sn + (n+1)H_{n+1} = \sum_{k=0}^{n} kH_{k+1} + \sum_{k=0}^{n} H_{k+1} $

## <center> Consider $ \sum_{k=0}^{n} kH_{k+1} + \sum_{k=0}^{n} H_{k+1} $

## $ \sum_{k=0}^{n} kH_{k+1} + \sum_{k=0}^{n} H_{k+1} $

## $ \sum_{k=0}^{n} (kH_k + \frac{k}{k+1}) + \sum_{k=0}^{n} H_{k+1} $

## $ \sum_{k=0}^{n} kH_k + \sum_{k=0}^{n}\frac{k}{k+1} + \sum_{k=0}^{n} H_{k+1} $

## $ S_n + \sum_{k=0}^{n}\frac{k}{k+1} + \sum_{k=0}^{n} H_{k+1} $

## $ Sn + (n+1)H_{n+1} = S_n + \sum_{k=0}^{n}\frac{k}{k+1} + \sum_{k=0}^{n} H_{k+1} $

## $ Sn - Sn + (n+1)H_{n+1} - \sum_{k=0}^{n}\frac{k}{k+1} = \sum_{k=0}^{n} H_{k+1} $

## $ (n+1)H_{n+1} - \sum_{k=0}^{n}\frac{k}{k+1} = \sum_{k=0}^{n} H_{k+1} $

## <center> Consider $ \sum_{k=0}^{n}\frac{k}{k+1} $

## $ \sum_{k=0}^{n}\frac{k}{k+1} $

## $ \sum_{k=0}^{n}(1 - \frac{1}{k+1}) $

## $ \sum_{k=0}^{n}1 - \sum_{k=0}^{n} \frac{1}{k+1} $

## $ n + 1 - \sum_{k=0}^{n} \frac{1}{k+1} $

## $ n + 1 - H_{n+1} $

```python
n = 19
print(n - sum([1/(k+1) for k in range(n+1)]))
print(n - harmonic_sum(n+1))
```

## $ (n+1)H_{n+1} - (n + 1 - H_{n+1}) = \sum_{k=0}^{n} H_{k+1} $

## $ nH_{n+1} + H_{n+1} - n - 1 + H_{n+1} = \sum_{k=0}^{n} H_{k+1} $

## $ nH_{n+1} - n - 1 + 2H_{n+1} = \sum_{k=0}^{n} (H_{k}+ \frac{1}{k+1}) $

## $ nH_{n+1} - n - 1 + 2H_{n+1} = \sum_{k=0}^{n} H_{k} + \sum_{k=0}^{n} \frac{1}{k+1} $

## $ nH_{n+1} - n - 1 + 2H_{n+1} = \sum_{k=0}^{n} H_{k} + H_{n+1} $

## $ nH_{n+1} - n - 1 + H_{n+1} = \sum_{k=0}^{n} H_{k} $

## $ H_{n+1}(n+1) - n - 1 = \sum_{k=0}^{n} H_{k} $

```python
n = 317

print(sum([harmonic_sum(k) for k in range(n+1)]))
closed_form = harmonic_sum(n+1)*(n+1) - n - 1
print(closed_form)
```

# <center> Closed form $ \sum_{k=0}^{n} H_{k} = H_{n+1}(n+1) - n - 1 $
