# <center> Josephus problem for case 2

## $$ 
J(1) = 1; \\
J(2n) = 2 J(n) - 1,  \text{ for } n \ge 1; \\
J(2n + 1) = 2 J(n) + 1,  \text{ for }  n \ge 1. 
$$

## $$ 
J(1) = a; \\
J(2n) = 2 J(n) - b,  \text{ for } n \ge 1; \\
J(2n + 1) = 2 J(n) + c,  \text{ for }  n \ge 1. 
$$

```python
def josephus_2(n:int):
    if n == 1:
        return 1
    
    if n < 1:
        return 0
    
    if n % 2 == 0:
        n = n//2
        return 2 * josephus_2(n) - 1
    
    n -= 1
    n //= 2
    return 2 * josephus_2(n) + 1
```

```python
n = 15
print(josephus_2(n))
```

## <center> $ R_n=n $ case

$
1 = a; \\
2\cdot n = 2\cdot n + b; \\
2\cdot n + 1= 2\cdot n + c.
$

### $ a=1, b=0, c=1 $

###  $ n = A(n) + C(n) $

###  $ C(n) = n - A(n) => n - 2^m = l $

## $ n = 2^m + l $

## <center> What is $ B(n)? $

### $ 1 = A(n) - B(n) - C(n) $

### $ B(n) = A(n) - C(n) - 1 $

### $ B(n) = 2^m - l - 1 $

# <center> $ \text{ Recall } A(n), B(n), C(n) $

## $$
A(n) = 2^m;\\
B(n) = 2^m - l - 1;\\
C(n) = l;\\
a = 1, b = -1, c = 1
$$

# <center> Solution

```python
def closed_josephus_2(n: int):
    if n < 1:
        return 0
    
    m = floor(log(n, 2))
    l = n - 2**m

    A = lambda n: 2**m
    B = lambda n: 2**m - l - 1
    C = lambda n: l
    
    return A(n) - B(n) + C(n)

n = 16123123213123567534123214211241214215125215
print(josephus_2(n))
print(closed_josephus_2(n))
```
