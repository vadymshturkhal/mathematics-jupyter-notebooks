# <center> Knuth Multiply Permutations


## <center> Example

# $$(acf) \cdot (bd) \cdot (abd) \cdot (ef)  = (acefb) $$

## $$ (acf) = \begin{pmatrix} 
a & c & f \\ 
c & f & a
\end{pmatrix} $$

# <center> 1

## $$ (acf) \cdot (bd)  = \begin{pmatrix} 
a & c & f \\ 
c & f & a
\end{pmatrix} \cdot
\begin{pmatrix} 
b & d \\ 
d & b
\end{pmatrix}  = \begin{pmatrix} 
a & c & f & b & d\\ 
c & f & a & d & b
\end{pmatrix} $$

# <center> 2

## $$ (acf) \cdot (bd) \cdot (abd) = \begin{pmatrix} 
a & c & f & b & d\\

\end{pmatrix} \cdot \begin{pmatrix} 
a & b & d\\ 
b & d & a
\end{pmatrix} $$


## $$ (acf) \cdot (bd) \cdot (abd) = \begin{pmatrix} 
a & c & f & b & d\\ 
c & f & a & d & b
\end{pmatrix} \cdot \begin{pmatrix} 
a & d & b\\ 
b & a & d
\end{pmatrix} $$


## $$ (acf) \cdot (bd) \cdot (abd) = \begin{pmatrix}

\end{pmatrix} $$


## $$ (acf) \cdot (bd) \cdot (abd) = \begin{pmatrix}
